26 "Track" files, with concatenated sounds.

```python
nfiles = 26
soundbankdir = 'Soundbank_WB/'

for n in range(1,nfiles+1):
    fname = "{:02d} Track {:d}.aif".format(n,n)
    if(os.path.exists(fname)):
        print('processing', fname)
        sound = AudioSegment.from_file(fname)
        print("sound dB =", sound.dBFS)
        chunks = split_on_silence(
                sound,
                # split on silences longer than 1000ms (1 sec)
                min_silence_len=500,
                # anything under -50 dBFS is considered silence
                silence_thresh=-50, 
                # keep 200 ms of leading/trailing silence
                keep_silence=100)
        print(len(chunks), 'individual sounds found')

        #Export each segment to a file
        for i,c in enumerate(chunks):
            filename = soundbankdir + "{:d}-{:d}.aif".format(n, i+1)
            print('exporting ', filename)
            c.export(filename)
```
