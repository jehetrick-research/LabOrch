# Renaming sound files in McGill Bank

In this notebook we name the files following Francois' schema.

The McGill Soundbank files have been separated from the original McGill WMA files, which had many sounds in one WMA file. The separated files have only one sound: instrument, pitch, performance, and dynamic.

However, the files are simply named NN-XX, where NN is an instrument number within the directory, and XX is a sound number for that instrument; *e.g.,* in the directory Soundbank_Str/, the file named `1-20.aif` is the sound of a violin (bowed) playing a C5, in the *ord*inary way.

These have to be renamed according to the fields in Francois XSLX files, which are in ForJim/.

Loop over file/instrument in each sub-bank.

```python
for f in ppsfilenames:
    ppsdf = pd.read_excel('ForJim-04-08-20/' + f)
    for i,s in enumerate(ppsdf.iloc[:,0]):
        tens = np.floor(float(s))
        soundno = 100*(float(s)-tens)
        soundfname = "{:.0f}-{:.0f}.aif".format(tens,soundno)
        newsoundfname = "{}-{}-{}-{}.aif".format(ppsdf.iloc[i,1],ppsdf.iloc[i,2],ppsdf.iloc[i,3],ppsdf.iloc[i,4])
        print(ppsdf.columns[0], "|", soundfname, "|", newsoundfname)
```

$\checkmark$

```python
from os import rename
from os.path import exists
```

```python
bankdir = 'Soundbank_PianoPercSax/'

for f in ppsfilenames:
    ppsdf = pd.read_excel('ForJim-04-08-20/' + f)
    for i,s in enumerate(ppsdf.iloc[:,0]):
        tens = np.floor(float(s))
        soundno = 100*(float(s)-tens)
        soundfname = "{:.0f}-{:.0f}.aif".format(tens,soundno)
        newsoundfname = "{}-{}-{}-{}.aif".format(ppsdf.iloc[i,1],ppsdf.iloc[i,2],ppsdf.iloc[i,3],ppsdf.iloc[i,4])
        print('renaming:', bankdir + soundfname, ' -> ', bankdir + newsoundfname)
        if(exists(bankdir + soundfname)):
            rename(bankdir + soundfname, bankdir + newsoundfname)
```
