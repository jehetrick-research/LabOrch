# Splitting the McGill sound files

from: https://stackoverflow.com/questions/37725416/pydub-combine-split-on-silence-with-minimum-length-file-size

```python
from pydub import AudioSegment
from pydub.silence import split_on_silence

sound = AudioSegment.from_file("../SoundBank_McGill/Wind and Brass Samples/17 Track 17.aif")

chunks = split_on_silence(
    sound,
    # split on silences longer than 1000ms (1 sec)
    min_silence_len=500,
    # anything under -50 dBFS is considered silence
    silence_thresh=-50, 
    # keep 200 ms of leading/trailing silence
    keep_silence=100
)
```

```python
sound.dBFS
```

Number of segments split

```python
len(chunks)
```

Play one of the segments

```python
chunks[10]
```

Export each segment to a file

```python
for i,c in enumerate(chunks):
    filename = "17-{:02d}.aif".format(i+1)
    print(filename)
    c.export(filename)
```

```python
for i in range(len(numsounds)):
    f = "../SoundBank_McGill/Wind and Brass Samples/{:02d} Track {:d}.aif".format(i+1,i+1)
    print(i, f, os.path.exists(f))
    sound = AudioSegment.from_file(f)
    chunks = split_on_silence(
        sound,
        min_silence_len=500,
        silence_thresh=-50, 
        keep_silence=100
    )
    
    print("split file {}: len(chunks)={} numsounds={}".format(i+1, len(chunks), numsounds[i]))
```
