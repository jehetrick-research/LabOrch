# Counting the number of Bank sounds and FFTs

The number of *sounds* (AIFF files) in  *WWBr-IsolatedSounds/* is smaller than the number of *FFT* files we have in *FFTBank/*.

Let's explore.

## Read in files

```python
import glob

ffts = glob.glob('../FFTBank/*.nbb')
print(len(ffts), 'files')
print('e.g.:')
ffts[:5]
```

```python
sounds = glob.glob('../WWBr-IsolatedSounds/*/*.aif')
print(len(sounds), 'files')
print('e.g.:')
sounds[:5]
```

## Get name stubs

```python
print('FFTs:', len(ffts))
print('sounds:', len(sounds))
```

```python
ffts = [f.split('\\')[1].split('.')[0] for f in ffts]
```

```python
sounds = [f.split('\\')[2].split('.')[0] for f in sounds]
```

```python
both = [f for f in ffts if f in sounds]
```

```python
len(both)
```

```python
setffts = set(ffts)
setsounds = set(sounds)
```

```python
len(setffts & setsounds)
```

```python
setsounds - (setffts & setsounds)
```

We have many more FFTs (**983**) than sounds (**407**).

However, we have FFTs for all sounds (**381**), except:
* all Bass Trombone (Btb)
* Trumpet $D^\#_6~~$ (tpt-ds6-ord-mf)

These are missing FFT (*.nbb*) files. But we could construct them from the sound files

It would be nicer to have *sound files* for all **983** FFT files.  
That is much harder, because it is difficult to reconstruct a sound from an FFT power spectrum.
